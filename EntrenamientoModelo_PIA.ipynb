```python
!pip install face_recognition
!pip install opencv-python
import face_recognition
import os
import numpy as np
import pickle
```

```python
known_encodings = []
known_names = []

# Carpeta de imagenes
dataset_path = "dataset"

# Recorrer archivos
for filename in os.listdir(dataset_path):

    # Ruta completa
    image_path = os.path.join(
        dataset_path,
        filename
    )

    try:

        # Cargar imagen
        image = face_recognition.load_image_file(
            image_path
        )

        # Obtener embeddings
        encodings = face_recognition.face_encodings(
            image
        )

        # Si no hay rostro, ignorar
        if len(encodings) == 0:

            print(f"No se detecto rostro en {filename}")
            continue

        # Primer rostro encontrado
        encoding = encodings[0]

        # Guardar embedding
        known_encodings.append(encoding)

        # Guardar nombre
        known_names.append("Ivan")

        print(f"{filename} registrada")

    except Exception as e:

        print(f"Error en {filename}: {e}")

print("Entrenamiento completado")
```

```python
test_image = face_recognition.load_image_file("test.jpeg")

test_encoding = face_recognition.face_encodings(test_image)[0]

results = face_recognition.compare_faces(
    known_encodings,
    test_encoding
)

print(results)
```

```python
face_distances = face_recognition.face_distance(
    known_encodings,
    test_encoding
)

print("\n==============================")
print(" GUIA DE DISTANCIAS FACIALES ")
print("==============================")

print(
    "0.20 - 0.35  -> Excelente"
)

print(
    "0.35 - 0.45  -> Muy confiable"
)

print(
    "0.45 - 0.55  -> Aceptable"
)

print(
    "0.55 - 0.65  -> Dudosa"
)

print(
    "> 0.65       -> Otra persona"
)

print("==============================")


for i, distancia in enumerate(face_distances):

    print(
        f"Imagen {i+1}: "
        f"Distancia = {distancia:.4f} "
    )

promedio = np.mean(face_distances)

print("\n===================")

print(
    f"Promedio: "
    f"{promedio:.4f}"
)

if promedio < 0.45:
    print("MUY CONFIABLE")

elif promedio < 0.60:
    print("ACEPTABLE")

else:
    print("DESCONOCIDO")
```

```python
data = {
    "encodings": known_encodings,
    "names": known_names
}

with open("rostros.pkl", "wb") as f:
    pickle.dump(data, f)
```
